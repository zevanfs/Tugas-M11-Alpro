```python
#Soal 3.1
def hitung_luas_persegi(sisi):
    hasil = sisi * sisi
    print(f"Luas persegi satuan {sisi} adalah: {hasil}")
    return hasil

luas1 = hitung_luas_persegi(5)
luas2 = hitung_luas_persegi(10)
luas3 = hitung_luas_persegi(15)

rata_rata_luas = (luas1 + luas2 + luas3) /3
print(f"\nRata-rata luas 3 persegi adalah {rata_rata_luas:.4f}")
```

```python
def konversi_nilai(nilai_angka):
    if nilai_angka >= 86:
       return "A"
    elif (nilai_angka >= 76) and (nilai_angka < 86):
        return "AB"
    elif (nilai_angka >= 66) and (nilai_angka < 76):
        return "B"
    else:
        return "C"

nilai_angka = int(input("Masukkan nilai anda:"))
nilai_angka = konversi_nilai(nilai_angka)
print(f"Predikat nilai anda adalah {nilai_angka}")
```
